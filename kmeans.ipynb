# K-Means

```python
import math
```

```python
data = [(1, 1), (2, 1), (1, 2), (6, 5), (7, 5), (6, 6)]

K = 2
means = [(1, 1), (2, 1)]
```

euclidean distance

```python
def dist(a, b):
    return math.sqrt((a[0] - b[0]) ** 2 + (a[1] - b[1]) ** 2)
```

```python
step = 0

while True:
    step += 1
    clusters = [[] for _ in range(K)]

    for point in data:
        nearest = min(range(K), key=lambda i: dist(point, means[i]))
        clusters[nearest].append(point)

    new_means = []
    for c in clusters:
        mx = sum(p[0] for p in c) / len(c)
        my = sum(p[1] for p in c) / len(c)
        new_means.append((mx, my))

    J = sum(dist(p, new_means[i]) ** 2 for i in range(K) for p in clusters[i])
    print(f"Iteration {step}: centers = {new_means}, J = {J:.3f}")

    if new_means == means:
        break

    means = new_means
```

```python
for i in range(K):
    print(f"Cluster {i + 1}: {clusters[i]}")
    print(f"Center: {means[i]}")
    print()
```
